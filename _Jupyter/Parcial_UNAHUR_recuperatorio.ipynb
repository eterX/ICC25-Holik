Es importante remarcar que todo estado puro $\psi$ siempre se puede reescribir en términos de su matriz densidad, $\rho_{\psi} = |\psi\rangle\langle\psi|$ y esta matriz es hermítica, sus autovalores son no negativos, y suma da uno. Se puede probar que el valor medio de un observable $A$ (representado por una matriz hermítica), siempre se puede escribir como:

$$\langle A \rangle = tr(\rho_{\psi}A)$$

Para estados puros se cumple que:

$$\langle A \rangle = tr(\rho_{\psi}A)=\langle \psi|A|\psi\rangle$$

Sin embargo, no todos los estados son puros, en el sentido de que no todo estado de un qubit se puede representar con un vector $|\psi\rangle$. En presencia de ruido (o de la interacción con un entorno), es posible que el qubit se encuentre en un estado mixto. Los estados mixtos no se pueden escribir en términos de estados puros (es decir, dado un $\rho_{m}$ mixto, no existe ningún $\psi$ tal que $\rho_{m}$ sea igual a $|\psi\rangle\langle\psi|$). #### Valores medios de un observable

 Usando las fórmulas del ejercicio anterior (es decir, que el valor medio de un observable $A$ para un sistema en el etsado $\rho$ viene dado por $tr(\rho A)$), calcule los valores medios de:

$$\sigma_{x}=  \begin{pmatrix}
          0 & 1  \\
          1 & 0 
          \end{pmatrix},$$

$$\sigma_{y}=  \begin{pmatrix}
          0 & -i  \\
          i & 0 
          \end{pmatrix}$$
y 

$$\sigma_{z}=  \begin{pmatrix}
          1 & 0  \\
          0 & -1 
          \end{pmatrix}$$

para los estados $\rho_{\psi}$ y $\rho_{m}$ del ejercicio anterior. Compare los resultados obtenidos. ¿Son $\rho_{\psi}$ y $\rho_{m}$ estados físicamente equivalentes?

- [x] respuesta:


```python
#entonrno


from qiskit_aer import AerSimulator
import qiskit as qk
import numpy as np
import sympy as sp
from IPython.display import display, Math

sigma_x = np.array([[0,1],[1,0]])
display(Math("$\sigma_x$"),sp.Matrix(sigma_x))
sigma_y = np.array([[0,-1j],[1j,0]])
display(Math("$\sigma_y$"),sp.Matrix(sigma_y))
sigma_z = np.array([[1,0],[0,-1]])
display(Math("$\sigma_z$"),sp.Matrix(sigma_z))

rho_psi = np.array([[0.5, 0.5],
                    [0.5, 0.5]])
display(Math(r"$\rho_{\psi}$"),sp.Matrix(rho_psi))
rho_m = np.array([[0.5, 0.0],
                  [0.0, 0.5]])
display(Math(r"$\rho_m$"),sp.Matrix(rho_m))
```

verficio $\langle\sigma_x\rangle$ 

```python
rho_psi_dot_sigma_x = rho_psi @ sigma_x 
sigma_x_medio_psi = np.trace(rho_psi_dot_sigma_x)
display(Math(r"$\langle\sigma_x\rangle$"),sp.simplify(sigma_x_medio_psi))
rho_m_dot_sigma_x = rho_m @ sigma_x 
sigma_x_medio_m = np.trace(sigma_x)
display(Math(r"$\langle\sigma_x\rangle$"),sp.simplify(sigma_x_medio_m))
if np.allclose(sigma_x_medio_psi,sigma_x_medio_m):
    print(f"sigma_x_medio, de ambos {sigma_x_medio_psi} y {sigma_x_medio_m} son iguales")
else:
    print(f"sigma_x_medio, de ambos {sigma_x_medio_psi} y {sigma_x_medio_m} son distintos")
```

verficio $\langle\sigma_y\rangle$ 

```python
rho_psi_dot_sigma_y = rho_psi @ sigma_y 
sigma_y_medio_psi = np.trace(rho_psi_dot_sigma_y)
display(Math(r"$\langle\sigma_y\rangle$"),sp.simplify(sigma_y_medio_psi))
rho_m_dot_sigma_y = rho_m @ sigma_y 
sigma_y_medio_m = np.trace(sigma_y)
display(Math(r"$\langle\sigma_y\rangle$"),sp.simplify(sigma_y_medio_m))
if np.allclose(sigma_y_medio_psi,sigma_y_medio_m):
    print(f"sigma_y_medio, de ambos {sigma_y_medio_psi} y {sigma_y_medio_m} son iguales")
else:
    print(f"sigma_y_medio, de ambos {sigma_y_medio_psi} y {sigma_y_medio_m} son distintos")

```
