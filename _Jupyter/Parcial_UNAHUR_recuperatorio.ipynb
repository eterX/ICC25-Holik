## 1 Estados puros vs estados mixtos

Suponga que un qubit es preparado en el estado:

$$|\psi\rangle = \frac{1}{\sqrt{2}}(|0\rangle + |1\rangle)$$

Ese mismo estado se puede representar también usando su \textit{matriz densidad} asociada:

$$\rho_{\psi} = |\psi\rangle\langle\psi|$$

Caclcule explícitamente la matriz $\rho$ y: 

(i) Calcule los autovalores de $\rho_{\psi}$. Comprueba que no son negativos.

(ii) Compruebe que la suma de los autovalores de $\rho_{\psi}$ es igual a uno.

(iii) Compruebe que $\rho_{\psi}$ es Hermítica.

Es importante remarcar que todo estado puro $\psi$ siempre se puede reescribir en términos de su matriz densidad, $\rho_{\psi} = |\psi\rangle\langle\psi|$ y esta matriz es hermítica, sus autovalores son no negativos, y suma da uno. Se puede probar que el valor medio de un observable $A$ (representado por una matriz hermítica), siempre se puede escribir como:

$$\langle A \rangle = tr(\rho_{\psi}A)$$

Para estados puros se cumple que:

$$\langle A \rangle = tr(\rho_{\psi}A)=\langle \psi|A|\psi\rangle$$

Sin embargo, no todos los estados son puros, en el sentido de que no todo estado de un qubit se puede representar con un vector $|\psi\rangle$. En presencia de ruido (o de la interacción con un entorno), es posible que el qubit se encuentre en un estado mixto. Los estados mixtos no se pueden escribir en términos de estados puros (es decir, dado un $\rho_{m}$ mixto, no existe ningún $\psi$ tal que $\rho_{m}$ sea igual a $|\psi\rangle\langle\psi|$). Considere el estado representado por la siguiente matriz densidad:

$$\rho_{m}=  \begin{pmatrix}
          \frac{1}{2} & 0  \\
          0 & \frac{1}{2} 
          \end{pmatrix}$$

(a) Calcule sus autovalores. Compruebe que son no negativos y que suman uno.

(b) Compruebe que $\rho_{m}$ es hermítica.

Dado un estado $\phi$ y un estado $\psi$, la probabilidad de transición venía dada por $|\langle\phi|\psi\rangle|^{2}$. Por eso, la probabilidad $P_{0}^{\psi}$ de observar el resultado $0$ dado el estado $|\psi\rangle$ viene dada por $|\langle 0|\psi\rangle|^{2}$. Si $\rho_{0}=|0\rangle\langle 0|$ y $\rho_{1}=|1\rangle\langle1|$, compruebe que:

$$P_{0}^{\psi} =tr(\rho_{0}\rho_{\psi})$$

y

$$P_{1}^{\psi} =tr(\rho_{1}\rho_{\psi})$$

Compare con:

$$P_{0}^{\rho_{m}} =tr(\rho_{0}\rho_{m})$$

y

$$P_{1}^{\rho_{m}} =tr(\rho_{1}\rho_{m})$$

Compruebe que 

$$\rho_{m} = \frac{1}{2}\rho_{0}+\frac{1}{2}\rho_{1}$$
